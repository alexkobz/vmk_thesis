```python
import pandas as pd

from src.utils import ProcessedCsvReader

df = ProcessedCsvReader().read()
```

```python
tmp = df[
    [col for col in df.columns if col.startswith('lr__sm__')] +
    [col for col in df.columns if col.startswith('cat__')] +
    [col for col in df.columns if col.startswith('mm__')] +
    [col for col in df.columns if col.startswith('st__')]
].fillna(0)
corr = tmp.corr()
```

```python
import numpy as np
# порог
threshold = 0.5
# верхний треугольник матрицы
upper = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)
# колонки, которые надо удалить
to_drop = [
    column
    for column in upper.columns
    if any(upper[column] > threshold)
]
# итоговые некоррелированные колонки
selected_cols = [
    col for col in tmp.columns
    if col not in to_drop
]
print(selected_cols)
```
