```python
from ngboost.distns import Normal
from ngboost.learners import default_linear_learner
from sktime.forecasting.compose import YfromX
from ngboost import NGBRegressor
from src.training.read_data import CsvReader
from src.training.run_cfg import ForecastConfig
from src.training.train import main
```

```python
ngboost_ridge = ForecastConfig(
    model=YfromX(
        estimator=NGBRegressor(
            Base=default_linear_learner,
            Dist=Normal,
            n_estimators=100,
            learning_rate=0.01,
            random_state=42,
            verbose=True,
        ),
        pooling="local",
    ),
    run_name="ngboost_ridge",
    data_reader=CsvReader().read_data,
)

main(ngboost_ridge)
```
